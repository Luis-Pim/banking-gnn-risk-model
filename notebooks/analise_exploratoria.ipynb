```python
display(
    clientes.groupBy("segmento")
    .agg(
        F.count("*").alias("clientes"),
        F.round(F.avg("judicializou")*100,2).alias("taxa_jud")
    )
)
```

```python
clientes = clientes.withColumn(
    "faixa_idade",
    F.when(F.col("idade") < 30, "18-29")
     .when(F.col("idade") < 45, "30-44")
     .when(F.col("idade") < 60, "45-59")
     .otherwise("60+")
)


display(
    clientes.groupBy("faixa_idade")
    .agg(
        F.count("*").alias("clientes"),
        F.round(F.avg("judicializou")*100,2).alias("taxa_jud")
    )
)
```
