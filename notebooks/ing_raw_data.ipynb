### VALIDATION

```python
df = spark.table("clientes_raw")
edges = spark.table("edges_raw")
display(df.limit(10))
display(edges.limit(10))
```

```python
print("Clientes:", df.count())
print("Colunas clientes:", len(df.columns))
print("Edges:", edges.count())
```

```python
from pyspark.sql import functions as F

nulls_clientes = df.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df.columns
])

display(nulls_clientes)
```

```python
display(
    df
    .groupBy("judicializou")
    .agg(F.count("*").alias("qtd"))
    .withColumn("perc", F.round(F.col("qtd") / df.count() * 100, 2))
)
```

```python
display(
    edges
    .groupBy("edge_type")
    .agg(F.count("*").alias("qtd"))
    .orderBy(F.desc("qtd"))
)
```
